````python
best_model = None



#################################################################################
# TODO: Tune hyperparameters using the validation set. Store your best trained  #
# model in best_model.                                                          #
#                                                                               #
# To help debug your network, it may help to use visualizations similar to the  #
# ones we used above; these visualizations will have significant qualitative    #
# differences from the ones we saw above for the poorly tuned network.          #
#                                                                               #
# Tweaking hyperparameters by hand can be fun, but you might find it useful to  #
# write code to sweep through possible combinations of hyperparameters          #
# automatically like we did on thexs previous exercises.                          #
#################################################################################
hidden_sizes   = [64, 128]
learning_rates = [1e-3, 1.5e-3]
regularizations = [0.0, 1e-4]

best_val_acc = -1

for hs in hidden_sizes:
    for lr in learning_rates:
        for reg in regularizations:
            print('=== hidden=%d, lr=%e, reg=%e ===' % (hs, lr, reg))

            # 1) 每个组合新建模型，避免参数热启动
            model = TwoLayerNet(input_size, hs, num_classes,
                                weight_scale=1e-3, reg=reg)

            # 2) 每个组合新建 Solver
            solver = Solver(model, data,
                            update_rule='sgd',
                            optim_config={'learning_rate': lr},
                            lr_decay=0.95,
                            num_epochs=10, batch_size=100,
                            print_every=200, verbose=False)
            solver.train()

            # 3) 用验证集比较（solver 训练时已经记录了）
            val_acc = solver.best_val_acc
            print('val acc: %f' % val_acc)

            # 4) 更新最优
            if val_acc > best_val_acc:
                best_val_acc = val_acc
                best_model = model

print('best val acc: %f' % best_val_acc)
################################################################################
#                              END OF YOUR CODE                                #
################################################################################
````

在更大的数据集上训练 —— 正确
更多数据让模型看到更多变化，更难“记住”训练集，能显著缓解过拟合。这是最根本的办法（数据增强也算这一类）。

增加更多隐藏单元 —— 错误

增大正则化强度 —— 正确
L2 正则（或 dropout 等）会惩罚大权重，限制模型的有效复杂度，迫使它学更平滑、更通用的特征，从而减小训练/测试差距。注意别调得太大，否则会变成欠拟合，两者都低。
